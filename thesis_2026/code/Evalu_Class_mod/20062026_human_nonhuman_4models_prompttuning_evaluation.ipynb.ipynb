```python
# ============================================================
# CELL 1: Imports
# ============================================================
import os
import shutil
import torch
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm
from PIL import Image, UnidentifiedImageError
from transformers import CLIPProcessor, CLIPModel
import open_clip
import mobileclip
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix, accuracy_score, classification_report,
    precision_recall_fscore_support
)
from sklearn.preprocessing import LabelEncoder

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Gerät: {device}")

categories = ["human", "non-human"]
VALID_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}


# ============================================================
# CELL 2: Modelle definieren + Lade-/Encode-Funktionen
# ============================================================
MOBILECLIP_WEIGHTS = r"C:\Users\Lukas\masterthesis\thesis_2026\data\weights\mobileclip_weights\mobileclip_s1.pt"

MODEL_CONFIGS = [
    {"name": "CLIP-ViT-B32",   "type": "hf",        "model_id": "openai/clip-vit-base-patch32", "batch_size": 64},
    {"name": "CLIP-ViT-L14",   "type": "hf",        "model_id": "openai/clip-vit-large-patch14", "batch_size": 32},
    {"name": "EVA01-CLIP-g14", "type": "open_clip", "model_id": "EVA01-g-14", "pretrained": "laion400m_s11b_b41k", "batch_size": 16},
    {"name": "MobileCLIP-S1",  "type": "mobileclip", "model_id": "mobileclip_s1", "pretrained": MOBILECLIP_WEIGHTS, "batch_size": 32},
]

def load_model(config):
    if config["type"] == "hf":
        processor = CLIPProcessor.from_pretrained(config["model_id"])
        model = CLIPModel.from_pretrained(config["model_id"]).to(device)
        model.eval()
        return {"type": "hf", "model": model, "processor": processor}

    elif config["type"] == "open_clip":
        model, _, preprocess = open_clip.create_model_and_transforms(
            config["model_id"], pretrained=config["pretrained"]
        )
        model = model.to(device)
        model.eval()
        tokenizer = open_clip.get_tokenizer(config["model_id"])
        return {"type": "open_clip", "model": model, "preprocess": preprocess, "tokenizer": tokenizer}

    elif config["type"] == "mobileclip":
        model, _, preprocess = mobileclip.create_model_and_transforms(
            config["model_id"], pretrained=config["pretrained"], device=device
        )
        model.eval()
        tokenizer = mobileclip.get_tokenizer(config["model_id"])
        return {"type": "mobileclip", "model": model, "preprocess": preprocess, "tokenizer": tokenizer}

    else:
        raise ValueError(f"Unbekannter Modelltyp: {config['type']}")

def encode_text(mb, texts):
    if mb["type"] == "hf":
        # truncation=True als Sicherheitsnetz gegen CLIP's 77-Token-Limit
        inputs = mb["processor"](text=texts, return_tensors="pt", padding=True, truncation=True).to(device)
        with torch.no_grad():
            text_out = mb["model"].get_text_features(**inputs)
        feats = text_out if isinstance(text_out, torch.Tensor) else text_out.pooler_output
    elif mb["type"] in ("open_clip", "mobileclip"):
        tokens = mb["tokenizer"](texts).to(device)
        with torch.no_grad():
            feats = mb["model"].encode_text(tokens)
    return feats / feats.norm(dim=-1, keepdim=True)

def encode_images(mb, imgs):
    if mb["type"] == "hf":
        inputs = mb["processor"](images=list(imgs), return_tensors="pt", padding=True).to(device)
        with torch.no_grad():
            image_out = mb["model"].get_image_features(**inputs)
        feats = image_out if isinstance(image_out, torch.Tensor) else image_out.pooler_output
    elif mb["type"] in ("open_clip", "mobileclip"):
        tensors = torch.stack([mb["preprocess"](img) for img in imgs]).to(device)
        with torch.no_grad():
            feats = mb["model"].encode_image(tensors)
    return feats / feats.norm(dim=-1, keepdim=True)


# ============================================================
# CELL 3: Prompts (Simple / Detailed / Mixed)
# Detailed-Prompts gekürzt auf < 77 Tokens (CLIP-Limit!)
# ============================================================
prompt_variants = {
    "simple": {
        "human":     "a photo of a person",
        "non-human": "a photo of a landscape or scene"
    },
    "detailed": {
        "human":     "This is a close-up photo of a person or group of people posing for "
                     "the camera. Even if there is a beautiful mountain, landscape, or "
                     "scenic view in the background, the person in the foreground is what "
                     "this photo is about, not the scenery behind them.",
        "non-human": "This is a wide landscape or scenic view where no person is posing "
                     "prominently in the foreground. The terrain, vegetation, wildlife, or "
                     "structures fill the frame; any people present are small, distant, or "
                     "incidental background details."
    },
    "mixed": {
        "human":     "a selfie or portrait dominated by a person in the foreground, "
                     "regardless of the scenic background behind them",
        "non-human": "a wide scenic landscape dominated by terrain, vegetation, or "
                     "structures, with no person posing in the foreground"
    }
}

PROMPT_CANDIDATES = ["simple", "detailed", "mixed"]


# ============================================================
# CELL 4: Stratified Split — Dev vs. Test (EINMALIG, für alle Modelle gleich)
# ============================================================
source_folder     = r"C:\Users\Lukas\masterthesis\thesis_2026\data\Model_Evaluation_Human_Non_Human\random_sample_images_human_non_human_model_eva"
source_label_csv  = r"C:\Users\Lukas\masterthesis\thesis_2026\data\Model_Evaluation_Human_Non_Human\photo_labels_human_non_human_20062026.csv"

base_split_dir    = r"C:\Users\Lukas\masterthesis\thesis_2026\data\Model_Evaluation_Human_Non_Human\split_eva"
dev_image_folder  = os.path.join(base_split_dir, "dev")
test_image_folder = os.path.join(base_split_dir, "test")

SPLIT_SEED   = 42
DEV_FRACTION = 0.35  # ~35% Dev, ~65% Test

df_labels   = pd.read_csv(source_label_csv)
labels_dict = dict(zip(df_labels["file_name"], df_labels["Main_focus"]))

all_paths_per_class = {cat: [] for cat in categories}
for cat in categories:
    folder = os.path.join(source_folder, cat)
    if not os.path.exists(folder):
        print(f"⚠ Ordner nicht gefunden: {folder}")
        continue
    for fname in sorted(os.listdir(folder)):
        ext = os.path.splitext(fname)[1].lower()
        if ext not in VALID_EXTENSIONS:
            continue
        fpath = os.path.join(folder, fname)
        if os.path.isfile(fpath) and fname in labels_dict:
            all_paths_per_class[cat].append(fpath)

for cat in categories:
    print(f"{cat}: {len(all_paths_per_class[cat])} Bilder gefunden")

dev_paths, test_paths = [], []
for cat in categories:
    paths = all_paths_per_class[cat]
    dev_p, test_p = train_test_split(
        paths, train_size=DEV_FRACTION, random_state=SPLIT_SEED, shuffle=True
    )
    dev_paths.extend(dev_p)
    test_paths.extend(test_p)
    print(f"{cat}: Dev={len(dev_p)} | Test={len(test_p)}")

for cat in categories:
    os.makedirs(os.path.join(dev_image_folder, cat), exist_ok=True)
    os.makedirs(os.path.join(test_image_folder, cat), exist_ok=True)

def copy_to_split(paths, target_root):
    for p in paths:
        fname = os.path.basename(p)
        label = labels_dict[fname]
        dest = os.path.join(target_root, label, fname)
        if not os.path.exists(dest):
            shutil.copy(p, dest)

copy_to_split(dev_paths, dev_image_folder)
copy_to_split(test_paths, test_image_folder)

dev_label_csv  = os.path.join(base_split_dir, "dev_labels.csv")
test_label_csv = os.path.join(base_split_dir, "test_labels.csv")

pd.DataFrame({
    "file_name": [os.path.basename(p) for p in dev_paths],
    "Main_focus": [labels_dict[os.path.basename(p)] for p in dev_paths]
}).to_csv(dev_label_csv, index=False)

pd.DataFrame({
    "file_name": [os.path.basename(p) for p in test_paths],
    "Main_focus": [labels_dict[os.path.basename(p)] for p in test_paths]
}).to_csv(test_label_csv, index=False)

print(f"\nDev-Set:  {dev_image_folder}")
print(f"Test-Set: {test_image_folder}")

df_dev_labels    = pd.read_csv(dev_label_csv)
dev_labels_dict  = dict(zip(df_dev_labels["file_name"], df_dev_labels["Main_focus"]))
df_test_labels   = pd.read_csv(test_label_csv)
test_labels_dict = dict(zip(df_test_labels["file_name"], df_test_labels["Main_focus"]))

dev_image_paths = []
for cat in categories:
    folder = os.path.join(dev_image_folder, cat)
    for fname in sorted(os.listdir(folder)):
        fpath = os.path.join(folder, fname)
        if os.path.isfile(fpath) and fname in dev_labels_dict:
            dev_image_paths.append(fpath)

test_image_paths = []
for cat in categories:
    folder = os.path.join(test_image_folder, cat)
    for fname in sorted(os.listdir(folder)):
        fpath = os.path.join(folder, fname)
        if os.path.isfile(fpath) and fname in test_labels_dict:
            test_image_paths.append(fpath)

print(f"Dev-Bilder gesamt: {len(dev_image_paths)} | Test-Bilder gesamt: {len(test_image_paths)}")


# ============================================================
# CELL 5: Hilfsfunktionen — Klassifikation
# ============================================================
def safe_open(path):
    try:
        return Image.open(path).convert("RGB")
    except (UnidentifiedImageError, OSError) as e:
        print(f"  ⚠ Skipped ({os.path.basename(path)}): {e}")
        return None

def classify(model_bundle, image_paths, category_texts, batch_size):
    text_features = encode_text(model_bundle, category_texts)
    results, confidences, valid_paths = [], [], []

    for i in tqdm(range(0, len(image_paths), batch_size), desc="Batches", leave=False):
        batch_paths = image_paths[i : i + batch_size]
        loaded = [(p, safe_open(p)) for p in batch_paths]
        loaded = [(p, img) for p, img in loaded if img is not None]
        if not loaded:
            continue

        paths_ok, imgs_ok = zip(*loaded)
        image_features = encode_images(model_bundle, imgs_ok)

        similarities = image_features @ text_features.T
        batch_preds  = similarities.argmax(dim=1).cpu().numpy()
        batch_conf   = similarities.max(dim=1).values.cpu().float().numpy()

        results.extend([categories[idx] for idx in batch_preds])
        confidences.extend(batch_conf)
        valid_paths.extend(paths_ok)

    return valid_paths, results, confidences


# ============================================================
# CELL 6: Hauptschleife — pro Modell: Prompt-Tuning (Dev) + Evaluation (Test)
# ============================================================
output_root = r"C:\Users\Lukas\masterthesis\thesis_2026\data\Model_Evaluation_Human_Non_Human_23062026\Eva_AllModels_results_split"
os.makedirs(output_root, exist_ok=True)

all_model_results = {}
all_prompt_comparisons = []  # sammelt Prompt-Vergleiche über ALLE Modelle hinweg

for config in MODEL_CONFIGS:
    model_name = config["name"]
    batch_size = config["batch_size"]
    print(f"\n{'='*60}\nModell: {model_name}\n{'='*60}")

    model_bundle = load_model(config)
    model_output_dir = os.path.join(output_root, model_name)
    os.makedirs(model_output_dir, exist_ok=True)

    # --- Schritt 1: Prompt-Vergleich auf dem Dev-Set ---
    prompt_results = {}
    prompt_comparison_rows = []  # nur für dieses Modell

    for variant_name in PROMPT_CANDIDATES:
        category_texts = [prompt_variants[variant_name][cat] for cat in categories]
        valid_paths, preds, _ = classify(model_bundle, dev_image_paths, category_texts, batch_size)
        true = [dev_labels_dict[os.path.basename(p)] for p in valid_paths]

        precision, recall, f1, _ = precision_recall_fscore_support(true, preds, labels=categories, zero_division=0)
        macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(true, preds, labels=categories, average="macro", zero_division=0)
        acc = accuracy_score(true, preds)

        prompt_results[variant_name] = {"f1": dict(zip(categories, f1)), "accuracy": acc, "macro_f1": macro_f1}
        print(f"  [{variant_name:8s}] Dev-Acc={acc:.4f} | Macro-F1={macro_f1:.4f}")

        row = {
            "Modell": model_name,
            "Prompt": variant_name,
            "Dev Accuracy": round(acc, 4),
            "Macro F1": round(macro_f1, 4),
            "F1 human": round(dict(zip(categories, f1))["human"], 4),
            "F1 non-human": round(dict(zip(categories, f1))["non-human"], 4),
        }
        prompt_comparison_rows.append(row)
        all_prompt_comparisons.append(row)

    # --- Detaillierte Prompt-Vergleichstabelle ausgeben + speichern ---
    df_prompt_comparison = pd.DataFrame(prompt_comparison_rows)
    print(df_prompt_comparison.to_string(index=False))
    prompt_comparison_csv = os.path.join(model_output_dir, f"{model_name}_prompt_comparison.csv")
    df_prompt_comparison.to_csv(prompt_comparison_csv, index=False)
    print(f"  Prompt-Vergleich gespeichert: {prompt_comparison_csv}")

    # --- Schritt 2: Mixed-Prompt-Set pro Klasse bestimmen ---
    mixed_prompts = {}
    for cat in categories:
        f1_scores = {v: prompt_results[v]["f1"][cat] for v in PROMPT_CANDIDATES}
        best_variant = max(f1_scores, key=f1_scores.get)
        mixed_prompts[cat] = prompt_variants[best_variant][cat]
    print(f"  Mixed-Set: { {cat: v[:40]+'...' if len(v)>40 else v for cat, v in mixed_prompts.items()} }")

    # --- Schritt 3: Finale Evaluation auf dem TEST-SET ---
    mixed_texts = [mixed_prompts[cat] for cat in categories]
    valid_paths, results, confidences = classify(model_bundle, test_image_paths, mixed_texts, batch_size)
    true = [test_labels_dict[os.path.basename(p)] for p in valid_paths]

    acc = accuracy_score(true, results)
    macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(true, results, labels=categories, average="macro", zero_division=0)
    precision, recall, f1, _ = precision_recall_fscore_support(true, results, labels=categories, zero_division=0)

    print(f"  >>> TEST-SET  Accuracy={acc:.4f} | Macro-P={macro_p:.4f} | Macro-R={macro_r:.4f} | Macro-F1={macro_f1:.4f}")

    # --- Schritt 4: Testbilder nach Vorhersage sortieren ---
    sorted_dirs = {cat: os.path.join(model_output_dir, "sorted", cat) for cat in categories}
    for d in sorted_dirs.values():
        os.makedirs(d, exist_ok=True)

    for path, pred in zip(valid_paths, results):
        shutil.copy(path, sorted_dirs[pred])

    # --- CSV speichern ---
    df_results = pd.DataFrame({
        "file_name": [os.path.basename(p) for p in valid_paths],
        "true_label": true,
        "pred_label": results,
        "confidence": confidences,
        "correct": [t == p for t, p in zip(true, results)]
    })
    output_csv = os.path.join(model_output_dir, f"{model_name}_test_predictions.csv")
    df_results.to_csv(output_csv, index=False)
    print(f"  Test-Predictions gespeichert: {output_csv}")
    print(df_results["pred_label"].value_counts())

    # --- Konfidenz visualisieren ---
    plt.figure(figsize=(8, 4))
    plt.hist(confidences, bins=30, alpha=0.7, color="steelblue")
    plt.xlabel("Cosine Similarity")
    plt.ylabel("Number of Images")
    plt.title(f"Confidence Distribution – {model_name} (Test-Set, Mixed Prompts)")
    plt.tight_layout()
    plt.savefig(os.path.join(model_output_dir, f"{model_name}_confidence.png"), dpi=150, bbox_inches='tight')
    plt.show()

    # --- Evaluation: Confusion Matrix + Classification Report ---
    le = LabelEncoder()
    le.classes_ = np.array(sorted(categories))
    cm = confusion_matrix(le.transform(true), le.transform(results))

    print()
    print(classification_report(true, results, target_names=sorted(categories)))

    pd.DataFrame(
        classification_report(true, results, target_names=sorted(categories), output_dict=True)
    ).transpose().to_csv(os.path.join(model_output_dir, f"{model_name}_metrics.csv"))

    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=le.classes_, yticklabels=le.classes_)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(f"Confusion Matrix – {model_name} (Test-Set, Accuracy: {acc:.2%})")
    plt.tight_layout()
    plt.savefig(os.path.join(model_output_dir, f"{model_name}_confusion_matrix.png"), dpi=150, bbox_inches='tight')
    plt.show()

    all_model_results[model_name] = {
        "mixed_prompts": mixed_prompts,
        "test_accuracy": acc,
        "test_macro_precision": macro_p,
        "test_macro_recall": macro_r,
        "test_macro_f1": macro_f1,
    }

    del model_bundle
    torch.cuda.empty_cache()


# ============================================================
# CELL 7: Gesamtvergleich aller Modelle (Test-Set) + Prompt-Vergleich gesamt
# ============================================================
print("\n" + "="*60)
print("GESAMTVERGLEICH — alle Modelle, jeweils mit eigenem Mixed-Prompt-Set")
print("="*60)

comparison_rows = []
for model_name, r in all_model_results.items():
    comparison_rows.append({
        "Modell": model_name,
        "Test Accuracy (%)": round(r["test_accuracy"] * 100, 1),
        "Macro Precision": round(r["test_macro_precision"], 3),
        "Macro Recall": round(r["test_macro_recall"], 3),
        "Macro F1": round(r["test_macro_f1"], 3),
    })

df_comparison = pd.DataFrame(comparison_rows).sort_values("Macro F1", ascending=False)
print(df_comparison.to_string(index=False))

comparison_csv = os.path.join(output_root, "model_comparison.csv")
df_comparison.to_csv(comparison_csv, index=False)
print(f"\nGespeichert: {comparison_csv}")

# --- Prompt-Vergleich über ALLE Modelle hinweg speichern ---
df_all_prompts = pd.DataFrame(all_prompt_comparisons)
all_prompts_csv = os.path.join(output_root, "all_models_prompt_comparison.csv")
df_all_prompts.to_csv(all_prompts_csv, index=False)
print(f"Gespeichert: {all_prompts_csv}")
print("\nPrompt-Vergleich (alle Modelle):")
print(df_all_prompts.to_string(index=False))
```
