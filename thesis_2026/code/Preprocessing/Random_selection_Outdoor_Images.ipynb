```python
# ============================================================
# Random Sample – 300 Outdoor Bilder für manuelle Klassifikation
# ============================================================
import os
import random
import pandas as pd
import shutil

# ── Pfade ────────────────────────────────────────────────────
IMAGE_FOLDER  = r'C:\Users\Lukas\masterthesis\thesis_2026\data\all_images_sorted\outdoor'
OUTPUT_CSV    = r'C:\Users\Lukas\masterthesis\thesis_2026\data\manual_classification_sample.csv'
OUTPUT_EXCEL  = r'C:\Users\Lukas\masterthesis\thesis_2026\data\manual_classification_sample.xlsx'
OUTPUT_FOLDER = r'C:\Users\Lukas\masterthesis\thesis_2026\data\manual_sample_images'

SEED        = 44
SAMPLE_SIZE = 300

# ── Bilder einlesen ──────────────────────────────────────────
SUPPORTED_FORMATS = ('.jpg', '.jpeg', '.png', '.JPG', '.JPEG', '.PNG')

all_images = [
    f for f in os.listdir(IMAGE_FOLDER)
    if f.endswith(SUPPORTED_FORMATS)
]

print(f'Gefundene Bilder gesamt: {len(all_images)}')

# ── Zufällige Stichprobe ─────────────────────────────────────
random.seed(SEED)
sample = random.sample(all_images, min(SAMPLE_SIZE, len(all_images)))
sample_sorted = sorted(sample)

print(f'Stichprobengröße:        {len(sample_sorted)}')
print(f'Seed:                    {SEED}')

# ── Bilder in neuen Ordner kopieren ──────────────────────────
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

for fname in sample_sorted:
    src = os.path.join(IMAGE_FOLDER, fname)
    dst = os.path.join(OUTPUT_FOLDER, fname)
    shutil.copy(src, dst)

print(f'\n✅ Bilder kopiert nach: {OUTPUT_FOLDER}')

# ── CSV + Excel erstellen ────────────────────────────────────
df = pd.DataFrame({
    'filename':     sample_sorted,
    'human_non_human': [''] * len(sample_sorted),  # ← human / non_human
})

df.to_csv(OUTPUT_CSV, index=False)
df.to_excel(OUTPUT_EXCEL, index=False)

print(f'✅ CSV gespeichert:   {OUTPUT_CSV}')
print(f'✅ Excel gespeichert: {OUTPUT_EXCEL}')
print()
print('=== Vorschau erste 5 Zeilen ===')
print(df.head().to_string())
```
