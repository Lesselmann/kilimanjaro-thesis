```python
# ============================================================
# Label-CSV erstellen aus Ordnerstruktur
# Erstellt photo_labels_nature_human.csv
# Augmentierte Bilder werden automatisch übersprungen
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd

# ── Pfade anpassen! ──────────────────────────────────────────
base_dir = '/content/drive/MyDrive/flickr_data_subset/subset/'

image_folders = {
    'indoor':  os.path.join(base_dir, 'indoor'),
    'outdoor': os.path.join(base_dir, 'outdoor')
}

output_csv = os.path.join(base_dir, 'photo_labels_indoor_outdoor_20042026.csv')

# ── Erlaubte Bildformate ──────────────────────────────────────
VALID_EXTENSIONS = ('.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff', '.webp')

# ── Labels sammeln ────────────────────────────────────────────
file_names = []
labels     = []
skipped    = 0

for label, folder in image_folders.items():
    if not os.path.exists(folder):
        print(f"⚠ Ordner nicht gefunden: {folder}")
        continue

    for fname in sorted(os.listdir(folder)):

        # Augmentierte Bilder überspringen
        if fname.startswith("augmented_"):
            skipped += 1
            continue

        # Nur Bilddateien
        if not fname.lower().endswith(VALID_EXTENSIONS):
            continue

        file_names.append(fname)
        labels.append(label)

# ── CSV speichern ─────────────────────────────────────────────
df = pd.DataFrame({
    'file_name':  file_names,
    'Main_focus': labels
})

df.to_csv(output_csv, index=False)

# ── Zusammenfassung ───────────────────────────────────────────
print(f"CSV gespeichert: {output_csv}")
print(f"Übersprungen (augmented_*): {skipped}")
print()
print("Verteilung:")
print(df['Main_focus'].value_counts())
print(f"\nGesamt: {len(df)} Bilder")
```
